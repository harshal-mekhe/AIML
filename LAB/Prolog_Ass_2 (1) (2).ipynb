**Theory**<br>
Prolog is a logic programming language used in Artificial Intelligence and knowledge-based systems.
A Prolog program consists mainly of facts, rules, and queries.
Facts represent known information, while rules define relationships between facts.
Queries are used to retrieve information and verify conditions from the knowledge base.
In this experiment, Prolog is used to represent and retrieve information about doctors, patients, diseases, and treatments.

**Algorithm**<br>
Start the program.
Define the doctors using the doctor predicate.
Define the patients using the patient predicate.
Define doctor specializations using the specialist predicate.
Define which doctors treat which patients using the treats predicate.
Define diseases of patients using the disease predicate.
Define rules for identifying:
Patients
Doctors
Treating doctors
Patient diseases
Specialist doctors

Execute different queries.

```python
% Facts

doctor(amit).
doctor(neha).
doctor(rahul).

patient(rohan).
patient(priya).
patient(karan).
patient(sneha).

specialist(amit,cardiology).
specialist(neha,neurology).
specialist(rahul,orthopedics).

treats(amit,rohan).
treats(amit,priya).
treats(neha,karan).
treats(rahul,sneha).

disease(rohan,heart_disease).
disease(priya,heart_disease).
disease(karan,migraine).
disease(sneha,fracture).

% Rules

is_patient(X) :-
    patient(X).

is_doctor(X) :-
    doctor(X).

treating_doctor(D,P) :-
    treats(D,P),
    doctor(D),
    patient(P).

patient_disease(P,D) :-
    disease(P,D).

specialist_doctor(D,S) :-
    specialist(D,S).

has_heart_disease(P) :-
    disease(P,heart_disease).

doctor_for_disease(D,P) :-
    treats(D,P),
    disease(P,_).

same_disease(P1,P2) :-
    disease(P1,D),
    disease(P2,D),
    P1 \= P2.
```
