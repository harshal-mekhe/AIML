## 1. Theoretical Background

### 1.1 Introduction to Classification

Classification is a supervised machine learning technique in which a model learns from labelled examples and predicts the category or class of new data. During training, the algorithm studies the relationship between input features and known output labels. It then uses the learned patterns to classify unseen records.

In this assignment, loan approval is a **binary classification** problem because the target variable has two possible classes:

- **Approved (Y / 1):** The loan application is approved.
- **Rejected (N / 0):** The loan application is rejected.

The input features include applicant income, co-applicant income, loan amount, credit history, education, employment status, and property area. The model uses these attributes to predict the loan status.

### 1.2 Support Vector Machine (SVM)

Support Vector Machine is a supervised learning algorithm commonly used for classification. It attempts to find a decision boundary, called a **hyperplane**, that separates observations belonging to different classes.

SVM chooses a boundary that maximizes the margin, which is the distance between the hyperplane and the nearest training observations from each class. These nearest observations are called **support vectors**. A larger margin can help the model generalize to unseen data.

For data that cannot be separated well using a straight boundary, SVM can use kernel functions such as linear, polynomial, and radial basis function (RBF) kernels to represent more complex decision boundaries.

**Advantages:**
- Effective in high-dimensional feature spaces.
- Can model linear and non-linear decision boundaries using suitable kernels.
- Uses support vectors to define the decision boundary.

**Limitations:**
- Choice of kernel and parameters can affect performance.
- Training can be expensive on large datasets.
- The resulting model can be less interpretable than a decision tree.

**Use in this assignment:** A linear SVM is trained after numerical scaling and categorical encoding. It predicts whether an applicant's loan should be approved or rejected.

### 1.3 Decision Tree Classifier

A Decision Tree is a supervised learning algorithm that makes predictions by splitting data into smaller groups based on feature values. It has a tree-like structure consisting of a root node, decision nodes, branches, and leaf nodes.

At each decision node, the algorithm selects a feature and a split that separates the classes. Common split criteria include **Gini impurity** and **entropy/information gain**. The process continues until a stopping condition is reached. A leaf node gives the predicted class.

For example, a tree may first split applicants based on credit history and then use income or loan amount in later decisions.

**Advantages:**
- Easy to understand and visualize.
- Can capture non-linear relationships.
- Does not require feature scaling.
- Can work with numerical and categorical features after suitable preprocessing.

**Limitations:**
- A deep tree can overfit the training data.
- Small changes in data can produce a different tree.
- A single tree may be less accurate than an ensemble of trees.

**Use in this assignment:** A Decision Tree classifier is trained with a maximum depth limit to reduce the risk of overfitting.

### 1.4 K-Nearest Neighbors (KNN)

K-Nearest Neighbors is a supervised, instance-based classification algorithm. It predicts the class of a new observation by finding the **K closest observations** in the training dataset.

The distance between observations is often calculated using Euclidean distance. For classification, the model looks at the labels of the K nearest neighbors and assigns the class that occurs most frequently among them.

For example, if K = 5 and three of the five nearest applicants have approved loans, KNN predicts the approved class.

**Advantages:**
- Simple to understand and implement.
- Does not require a complex training process.
- Can model decision boundaries that are not linear.

**Limitations:**
- Prediction can be slow for large training datasets.
- Results depend on the selected value of K and distance measure.
- Sensitive to feature scales, irrelevant features, and noisy data.

**Use in this assignment:** KNN is trained with K = 5. Standardization is important because features such as income and credit history have different numerical scales.

### 1.5 Naïve Bayes Classifier

Naïve Bayes is a probabilistic classification algorithm based on **Bayes' theorem**. It calculates the probability of a class given the observed input features and predicts the class with the highest posterior probability.

Bayes' theorem is:

\[
P(C \mid X) = \frac{P(X \mid C)P(C)}{P(X)}
\]

Where:
- \(P(C \mid X)\) is the posterior probability of class \(C\) given features \(X\).
- \(P(X \mid C)\) is the likelihood of observing features \(X\) given class \(C\).
- \(P(C)\) is the prior probability of class \(C\).
- \(P(X)\) is the probability of observing features \(X\).

The algorithm is called “naïve” because it assumes that features are conditionally independent given the class. This assumption simplifies probability calculations, although it may not always hold in real datasets.

**Advantages:**
- Fast to train and predict.
- Works well as a baseline for many classification tasks.
- Produces class probabilities.

**Limitations:**
- The independence assumption may be unrealistic when features are related.
- Probability estimates may be less accurate when the assumption is strongly violated.
- The choice of Naïve Bayes variant should match the feature distribution.

**Use in this assignment:** Gaussian Naïve Bayes is used. It models each transformed feature with a Gaussian (normal) distribution within each class.

### 1.6 Summary of the Four Algorithms

| Algorithm | Main idea | Important consideration |
|---|---|---|
| SVM | Finds a separating hyperplane with a large margin | Scaling and kernel selection |
| Decision Tree | Splits data using feature-based rules | Control tree depth to reduce overfitting |
| KNN | Predicts from the labels of nearby observations | Scaling and choice of K |
| Naïve Bayes | Uses Bayes' theorem and class probabilities | Conditional-independence assumption |

No algorithm is guaranteed to perform best on every dataset. Their actual performance must be measured on data that was not used for training.


## 2. Theory of Evaluation Metrics

After training the classification models, evaluation metrics are used to measure how well each model predicts loan approval status. The metrics in this assignment are calculated with **Approved (1)** treated as the positive class.

### 2.1 Confusion Matrix

A confusion matrix compares actual labels with predicted labels. For binary classification, it contains four outcomes:

| Term | Meaning in this assignment |
|---|---|
| True Positive (TP) | An actually approved application predicted as approved |
| True Negative (TN) | An actually rejected application predicted as rejected |
| False Positive (FP) | An actually rejected application predicted as approved |
| False Negative (FN) | An actually approved application predicted as rejected |

These four values are used to calculate several evaluation metrics.

### 2.2 Accuracy

Accuracy measures the proportion of all predictions that are correct.

\[
Accuracy = \frac{TP + TN}{TP + TN + FP + FN}
\]

A higher accuracy indicates that more predictions are correct overall. However, accuracy alone can be misleading when the classes are imbalanced.

### 2.3 Precision

Precision measures how many applications predicted as approved were actually approved.

\[
Precision = \frac{TP}{TP + FP}
\]

High precision means fewer rejected applications are incorrectly predicted as approved. In a lending context, this helps examine the risk of false approvals, although it does not by itself measure a borrower's future repayment ability.

### 2.4 Recall

Recall measures how many of the actually approved applications were correctly identified by the model.

\[
Recall = \frac{TP}{TP + FN}
\]

High recall means fewer actually approved applications are incorrectly predicted as rejected. This is useful when examining the risk of false rejections.

### 2.5 F1-Score

F1-score is the harmonic mean of precision and recall. It provides a single measure that balances both metrics.

\[
F1 = 2 \times \frac{Precision \times Recall}{Precision + Recall}
\]

A higher F1-score indicates a better balance between precision and recall. It is useful when both types of classification error matter.

### 2.6 Log Loss (Classification Loss)

Log loss, also called cross-entropy loss for binary classification, evaluates the quality of predicted probabilities. It penalizes predictions that assign a low probability to the actual class, with especially strong penalties for confident incorrect predictions.

For \(N\) observations:

\[
LogLoss = -\frac{1}{N}\sum_{i=1}^{N}
\left[y_i\log(p_i)+(1-y_i)\log(1-p_i)\right]
\]

Where:
- \(N\) is the number of test observations.
- \(y_i\) is the actual class (0 or 1).
- \(p_i\) is the predicted probability of the positive class (Approved).

**Lower log loss is better.** Unlike accuracy, log loss evaluates predicted probabilities, not just the final predicted class. In this notebook, probability predictions are obtained using each model's `predict_proba()` method.

### 2.7 Summary of Evaluation Metrics

| Metric | What it measures | Preferred result |
|---|---|---|
| Accuracy | Overall correctness | Higher |
| Precision | Correctness of predicted approvals | Higher |
| Recall | Coverage of actual approvals | Higher |
| F1-score | Balance of precision and recall | Higher |
| Log loss | Quality of predicted probabilities | Lower |

**Important:** In loan approval, false approvals and false rejections can have different consequences. Therefore, the best model should not be selected using accuracy alone. The evaluation should consider the business context, class balance, fairness, explainability, and the cost of different errors.


## 3. Import required libraries

## 7. Data preprocessing

The `Loan_ID` column is an identifier, so it is removed. The target is converted to numeric form (`Y` = 1, `N` = 0). Numerical missing values are imputed using the median; categorical missing values are imputed using the most frequent category. Categorical features are one-hot encoded, and numerical features are standardized.

Preprocessing is placed inside a scikit-learn pipeline so that imputers, encoders, and scalers are fitted on the training data only, reducing data leakage.


```python
# Remove identifier column if it exists
data = df.drop(columns=["Loan_ID"], errors="ignore").copy()

# Remove rows with missing target labels, if any
data = data.dropna(subset=["Loan_Status"])

X = data.drop(columns=["Loan_Status"])
y = data["Loan_Status"].map({"N": 0, "Y": 1})

# Keep only rows with valid target labels
valid_rows = y.notna()
X = X.loc[valid_rows]
y = y.loc[valid_rows].astype(int)

numerical_cols = X.select_dtypes(include=["number"]).columns.tolist()
categorical_cols = X.select_dtypes(exclude=["number"]).columns.tolist()

print("Numerical columns:", numerical_cols)
print("Categorical columns:", categorical_cols)
print("\\nTarget distribution (0 = Rejected, 1 = Approved):")
display(y.value_counts().sort_index())
```

```python
# Build preprocessing pipelines
numerical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer(transformers=[
    ("num", numerical_transformer, numerical_cols),
    ("cat", categorical_transformer, categorical_cols)
])

print("Preprocessing pipeline created successfully.")
```

## 8. Split data into training and testing sets
An 80:20 split is used. Stratification preserves the target class proportions in both subsets.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training records:", X_train.shape[0])
print("Testing records:", X_test.shape[0])
print("\\nTraining target distribution:")
display(y_train.value_counts(normalize=True).rename("Proportion"))
print("\\nTesting target distribution:")
display(y_test.value_counts(normalize=True).rename("Proportion"))
```

## 9. Build and train classification models
All models use the same preprocessing pipeline. SVM, KNN, and Naïve Bayes are trained on standardized numerical features and one-hot encoded categorical features.

```python
models = {
    "SVM": Pipeline(steps=[
        ("preprocessor", preprocessor),
        ("classifier", SVC(kernel="linear", probability=True, random_state=42))
    ]),
    "Decision Tree": Pipeline(steps=[
        ("preprocessor", preprocessor),
        ("classifier", DecisionTreeClassifier(max_depth=5, random_state=42))
    ]),
    "KNN": Pipeline(steps=[
        ("preprocessor", preprocessor),
        ("classifier", KNeighborsClassifier(n_neighbors=5))
    ]),
    "Naive Bayes": Pipeline(steps=[
        ("preprocessor", preprocessor),
        ("classifier", GaussianNB())
    ])
}

for name, model in models.items():
    model.fit(X_train, y_train)
    print(f"{name} trained successfully.")
```

## 10. Evaluate model performance

The positive class is **Approved (1)**. Precision, recall, and F1-score therefore describe performance for identifying approved applications. Log loss evaluates the quality of predicted probabilities; lower values are better.


```python
results = []
predictions = {}

for name, model in models.items():
    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, 1]
    predictions[name] = y_pred

    results.append({
        "Algorithm": name,
        "Accuracy": accuracy_score(y_test, y_pred),
        "Precision": precision_score(y_test, y_pred, zero_division=0),
        "Recall": recall_score(y_test, y_pred, zero_division=0),
        "F1-Score": f1_score(y_test, y_pred, zero_division=0),
        "Log Loss": log_loss(y_test, y_prob, labels=[0, 1])
    })

results_df = pd.DataFrame(results).set_index("Algorithm")
display(results_df.round(4))
```

### Classification reports

```python
for name, y_pred in predictions.items():
    print(f"\\n{name}")
    print(classification_report(
        y_test, y_pred,
        labels=[0, 1],
        target_names=["Rejected", "Approved"],
        zero_division=0
    ))
```

## 11. Confusion matrices

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 9))

for ax, (name, y_pred) in zip(axes.flatten(), predictions.items()):
    cm = confusion_matrix(y_test, y_pred, labels=[0, 1])
    sns.heatmap(
        cm, annot=True, fmt="d", cmap="Blues",
        xticklabels=["Rejected", "Approved"],
        yticklabels=["Rejected", "Approved"],
        ax=ax
    )
    ax.set_title(name)
    ax.set_xlabel("Predicted label")
    ax.set_ylabel("Actual label")

plt.tight_layout()
plt.show()
```

## 12. Comparative analysis

```python
# Compare accuracy, precision, recall, and F1-score
metric_columns = ["Accuracy", "Precision", "Recall", "F1-Score"]

results_df[metric_columns].plot(
    kind="bar", figsize=(11, 6), ylim=(0, 1.05)
)
plt.title("Classification Model Performance Comparison")
plt.ylabel("Score")
plt.xlabel("Algorithm")
plt.xticks(rotation=0)
plt.legend(loc="lower right")
plt.tight_layout()
plt.show()

# Compare log loss separately (lower is better)
plt.figure(figsize=(8, 4))
sns.barplot(x=results_df.index, y=results_df["Log Loss"])
plt.title("Log Loss Comparison (Lower is Better)")
plt.xlabel("Algorithm")
plt.ylabel("Log Loss")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

## 13. Identify the best model
The model with the highest F1-score is selected as the best balanced classifier in this experiment. This is one selection criterion, not a universal measure of lending suitability.

```python
best_algorithm = results_df["F1-Score"].idxmax()
best_row = results_df.loc[best_algorithm]

print("Best algorithm by F1-score:", best_algorithm)
print("Accuracy :", round(best_row["Accuracy"], 4))
print("Precision:", round(best_row["Precision"], 4))
print("Recall   :", round(best_row["Recall"], 4))
print("F1-score :", round(best_row["F1-Score"], 4))
print("Log loss :", round(best_row["Log Loss"], 4))
```

Use the actual metrics generated above to complete this discussion. Do not assume that one algorithm will always perform best.

- **SVM:** Can work well when classes have a clear separating boundary. It may require careful kernel and parameter selection.
