Vacuum Cleaner Agent Using Artificial Intelligence

**Aim:**      
To design and implement a simple AI-based hospital vacuum cleaner agent that can move between different hospital locations, detect dirty areas, clean them according to their priority, and manage its battery while performing cleaning tasks.

---

**Experimental Objective:**

* To represent a hospital as an environment.
* To implement a vacuum cleaner as an intelligent agent.
* To understand how the agent observes the hospital environment and selects appropriate cleaning actions.

---

**Expected Outcome:**

* Understand the basic architecture of an AI agent.
* Create a hospital environment using Python classes and dictionaries.
* Create an agent that can observe and modify its environment.
* Implement simple decision-making logic based on dirt level, priority, occupancy, and battery level.

---

**Concept:**

An AI agent is a system that observes its environment, makes decisions, and performs actions to achieve a particular goal.

In this experiment:

**Environment → Hospital**

1. **Agent →** Vacuum Cleaner
2. **Goal →** Clean dirty hospital locations
3. **Sensor →** Detect the dirt level of locations
4. **Action →** Move between locations and clean dirty areas
5. **Performance →** Cleaning score and efficient battery usage

```python
import random

class HospitalEnvironment:

    def __init__(self):
        self.rooms = {
            "Emergency Ward": {
                "dirt": random.randint(20, 90),
                "priority": 5,
                "occupied": random.choice([True, False])
            },

            "Operation Theater": {
                "dirt": random.randint(20, 90),
                "priority": 5,
                "occupied": random.choice([True, False])
            },

            "Patient Ward": {
                "dirt": random.randint(20, 90),
                "priority": 4,
                "occupied": random.choice([True, False])
            },

            "ICU": {
                "dirt": random.randint(20, 90),
                "priority": 5,
                "occupied": random.choice([True, False])
            },

            "Reception": {
                "dirt": random.randint(20, 90),
                "priority": 2,
                "occupied": random.choice([True, False])
            },

            "Corridor": {
                "dirt": random.randint(20, 90),
                "priority": 3,
                "occupied": False
            }
        }

class HospitalVacuumAgent:

    def __init__(self, environment):
        self.environment = environment
        self.currentLocation = "Corridor"
        self.battery = 100
        self.score = 0

    def display_environment(self):
        print("\nHospital Environment")
        print("----------------------------")

        for room, data in self.environment.rooms.items():
            print(
                room,
                "--> Dirt:", data["dirt"],
                ", Priority:", data["priority"],
                ", Occupied:", data["occupied"]
            )

    def calculate_score(self, room):
        data = self.environment.rooms[room]

        if data["occupied"]:
            return -1

        return data["dirt"] + (data["priority"] * 10)

    def select_room(self):
        best_room = None
        best_score = -1

        for room, data in self.environment.rooms.items():

            if data["dirt"] > 0 and not data["occupied"]:

                room_score = self.calculate_score(room)

                if room_score > best_score:
                    best_score = room_score
                    best_room = room

        return best_room

    def move_to(self, room):
        print("\nMoving from", self.currentLocation, "to", room)

        self.currentLocation = room
        self.battery -= 10
        self.score -= 1

    def clean_room(self, room):
        data = self.environment.rooms[room]

        print("Vacuum is checking", room)

        if data["occupied"]:
            print(room, "is occupied. Skipping...")
            return

        if data["dirt"] > 0:
            print(room, "is Dirty")
            print("Dirt Level:", data["dirt"])

            data["dirt"] = 0
            self.battery -= 15
            self.score += 10

            print(room, "has been Cleaned")

        else:
            print(room, "is Already Clean")

    def recharge(self):
        print("\nBattery is Low")
        print("Vacuum is going to Charging Station")

        self.currentLocation = "Charging Station"
        self.battery = 100

        print("Battery Charged to 100%")

    def clean_hospital(self):
        print("\nVacuum starts cleaning the hospital")
        print("-----------------------------------")

        while True:

            dirty_rooms = [
                room
                for room, data in self.environment.rooms.items()
                if data["dirt"] > 0
            ]

            if not dirty_rooms:
                break

            if self.battery < 30:
                self.recharge()

            room = self.select_room()

            if room is None:
                print("\nNo available unoccupied dirty room.")
                break

            self.move_to(room)
            self.clean_room(room)

            print("Battery:", self.battery)
            print("Score:", self.score)

    def final_report(self):
        print("\nFinal Hospital Environment")
        print("----------------------------")

        for room, data in self.environment.rooms.items():
            print(room, "| Dirt:", data["dirt"])

        print("\nFinal Location:", self.currentLocation)
        print("Final Battery:", self.battery)
        print("Performance Score:", self.score)

# Create hospital environment
environment = HospitalEnvironment()

# Create vacuum cleaner agent
vacuum = HospitalVacuumAgent(environment)

# Display initial environment
vacuum.display_environment()

# Start cleaning
vacuum.clean_hospital()

# Display final report
vacuum.final_report()
```

**Conclusion:** The Hospital Vacuum Cleaner Agent was successfully implemented using Python and basic Artificial Intelligence concepts.

The system models a hospital environment containing multiple locations with different dirt levels, priorities, and occupancy conditions. The vacuum cleaner observes the environment and selects a suitable location for cleaning based on these conditions instead of simply following a fixed sequence.

Thus, this experiment demonstrates the basic working of an intelligent agent and provides a simple foundation for developing more advanced autonomous cleaning systems for hospitals, colleges, offices, shopping malls, and other large environments.
